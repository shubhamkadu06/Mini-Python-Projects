```python
import sys

class ATM:
    def __init__(self, user_pin="1234", initial_balance=5000.0):
        # Initialize the ATM with a default PIN and starting balance
        self.pin = user_pin
        self.balance = initial_balance
        self.attempts = 3

    def check_pin(self):
        """Validates the user's PIN before allowing access."""
        print("=== Welcome to the Python Virtual ATM ===")
        while self.attempts > 0:
            entered_pin = input("Please enter your 4-digit PIN: ").strip()
            
            if entered_pin == self.pin:
                print("\nVerification successful! Welcome to your account.")
                return True
            else:
                self.attempts -= 1
                print(f"Incorrect PIN. Attempts remaining: {self.attempts}")
        
        print("\nCard locked due to too many failed attempts. Please contact your bank.")
        return False

    def show_menu(self):
        """Displays the main transaction menu."""
        while True:
            print("\n" + "="*30)
            print("          ATM MENU          ")
            print("="*30)
            print("1. Check Balance")
            print("2. Deposit Money")
            print("3. Withdraw Money")
            print("4. Change PIN")
            print("5. Exit")
            print("="*30)
            
            choice = input("Select an option (1-5): ").strip()
            
            if choice == '1':
                self.check_balance()
            elif choice == '2':
                self.deposit()
            elif choice == '3':
                self.withdraw()
            elif choice == '4':
                self.change_pin()
            elif choice == '5':
                print("\nThank you for using our ATM. Goodbye!")
                sys.exit()
            else:
                print("Invalid selection. Please choose a valid option from 1 to 5.")

    def check_balance(self):
        """Displays current available balance."""
        print(f"\nYour current balance is: ${self.balance:,.2f}")

    def deposit(self):
        """Handles cash deposit tracking."""
        try:
            amount = float(input("\nEnter the amount to deposit: $"))
            if amount > 0:
                self.balance += amount
                print(f"Successfully deposited ${amount:,.2f}.")
                self.check_balance()
            else:
                print("Transaction failed: Deposit amount must be greater than zero.")
        except ValueError:
            print("Invalid input. Please enter a valid numerical amount.")

    def withdraw(self):
        """Handles cash withdrawals with error handling."""
        try:
            amount = float(input("\nEnter the amount to withdraw: $"))
            if amount <= 0:
                print("Transaction failed: Withdrawal amount must be greater than zero.")
            elif amount > self.balance:
                print(f"Transaction declined: Insufficient funds. Available: ${self.balance:,.2f}")
            else:
                self.balance -= amount
                print(f"Successfully withdrew ${amount:,.2f}.")
                self.check_balance()
        except ValueError:
            print("Invalid input. Please enter a valid numerical amount.")

    def change_pin(self):
        """Allows users to safely change their login security code."""
        old_pin = input("\nEnter your current PIN to confirm: ").strip()
        if old_pin == self.pin:
            new_pin = input("Enter your new 4-digit PIN: ").strip()
            # Basic validation to ensure it's a 4-digit number
            if len(new_pin) == 4 and new_pin.isdigit():
                self.pin = new_pin
                print("PIN updated successfully!")
            else:
                print("Failed: PIN must be exactly 4 digits long and numerical.")
        else:
            print("Incorrect current PIN. Security change aborted.")

# Run the program
if __name__ == "__main__":
    # Create an ATM instance with a default PIN '1234' and a starting balance of $5,000
    my_atm = ATM(user_pin="1234", initial_balance=5000.0)
    
    # Authenticate user first, then launch menu if successful
    if my_atm.check_pin():
        my_atm.show_menu()
```
